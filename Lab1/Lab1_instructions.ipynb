Let's inspect the files directly:


```python
# View app.py content
with open('app.py', 'r') as f:
    print(f.read())
```

```python
# View Dockerfile content
with open('Dockerfile', 'r') as f:
    print(f.read())
```

### 🔍 Understanding the Dockerfile:
1. `FROM python:3.9.1`: Base image containing Python 3.9 runtime.
2. `ADD . /python-flask`: Copies application files into the container.
3. `WORKDIR /python-flask`: Sets default working directory inside the container.
4. `RUN pip install flask`: Installs required Python dependencies.
5. `EXPOSE 5000`: Informs Docker that container listens on port 5000.
6. `ENTRYPOINT [ "python", "app.py" ]`: Default command executed when container starts.


---
## ☘️ Step 3: Build the Docker Image
Build the container image with the tag `flask-app:v1`. The `.` tells Docker to look for the `Dockerfile` and build context in the current directory.


```python
!docker build -t flask-app:v1 .
```

---
## ☘️ Step 4: Verify the Built Image
List local Docker images to verify that `flask-app:v1` was created successfully.


```python
!docker images flask-app:v1
```

---
## ☘️ Step 5: Tag the Image for Docker Hub
To push an image to Docker Hub, tag it with your Docker Hub username.

> **Note:** Replace `yourdockerhubuser` below with your actual Docker Hub username.


```python
# Set your Docker Hub username
DOCKERHUB_USER = "yourdockerhubuser"

!docker tag flask-app:v1 {DOCKERHUB_USER}/flask-app:v1
!docker images | grep flask-app
```

## ☘️ Step 8: Run a Container from the Image
Launch a detached (`-d`) container named `flask-demo` mapping host port `5000` to container port `5000`.


```python
!docker run -d -p 5000:5000 --name flask-demo flask-app:v1
```

---
## ☘️ Step 9: Verify Running Container
Inspect running containers to ensure `flask-demo` status is `Up`.


```python
!docker ps --filter "name=flask-demo"
```

---
## ☘️ Step 10: Check Container Logs
View stdout/stderr generated by the Flask application inside the container.


```python
!docker logs flask-demo
```

---
## ☘️ Step 11: Access the Flask App
Test the HTTP endpoint running inside the container at `http://localhost:5000/`.


```python
import urllib.request

try:
    with urllib.request.urlopen("http://localhost:5000/", timeout=5) as response:
        print("Status Code:", response.getcode())
        print("Response Body:", response.read().decode('utf-8'))
except Exception as e:
    print("Could not reach endpoint locally:", e)
```

---
## ☘️ Step 12: Execute Commands Inside the Container
Run commands inside the running container to inspect files and processes.


```python
# List files inside container workdir
!docker exec flask-demo ls -la /python-flask
```

---
## ☘️ Step 13: Cleanup
Stop and remove the `flask-demo` container.


```python
!docker stop flask-demo
!docker rm flask-demo
```

```python
# Verify container is removed
!docker ps -a --filter "name=flask-demo"
```
